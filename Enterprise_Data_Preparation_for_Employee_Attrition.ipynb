```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

print("Libraries imported successfully!")
```

```python
from google.colab import files
import pandas as pd

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Dataset imported successfully!")
print("File name:", file_name)
print("Dataset shape:", df.shape)

df.head()
```

```python
# Dataset information
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all").T)
```

```python
# Check missing values
missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
display(missing_values)

# Total missing values
total_missing = df.isnull().sum().sum()

print("\nTotal Missing Values:", total_missing)

# Missing value percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("\nMissing Value Percentage:")
display(missing_percentage)
```

```python
# Check duplicate records
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)

# Remove duplicate records
if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print("Duplicate records removed successfully.")
else:
    print("No duplicate records found.")

print("\nDataset shape after duplicate handling:", df.shape)
```

```python
# Select numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(numeric_columns.tolist())

# Calculate outliers using IQR
outlier_summary = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

print("\nOutlier Detection Summary:")
display(outlier_df)
```

```python
# Create a copy before outlier treatment
df_clean = df.copy()

# Handle outliers using IQR capping
for col in numeric_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Cap extreme values
    df_clean[col] = df_clean[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Outlier treatment completed using IQR capping.")
print("Original shape:", df.shape)
print("After outlier treatment:", df_clean.shape)
```

```python
# Create a working copy
df_features = df_clean.copy()

# Display columns to confirm available HR fields
print("Available columns:")
print(df_features.columns.tolist())
```

```python
# Create a copy for feature engineering
df_features = df_clean.copy()

# 1. Salary per year
if "MonthlySalary" in df_features.columns:
    df_features["AnnualSalary"] = df_features["MonthlySalary"] * 12

# 2. Salary per year of experience
if "MonthlySalary" in df_features.columns and "YearsWithCompany" in df_features.columns:
    df_features["SalaryPerYearExperience"] = (
        df_features["MonthlySalary"] /
        (df_features["YearsWithCompany"] + 1)
    )

# 3. Leave rate
if "LeavesTaken" in df_features.columns and "YearsWithCompany" in df_features.columns:
    df_features["LeaveRate"] = (
        df_features["LeavesTaken"] /
        (df_features["YearsWithCompany"] + 1)
    )

# 4. Training intensity
if "TrainingHours" in df_features.columns and "YearsWithCompany" in df_features.columns:
    df_features["TrainingIntensity"] = (
        df_features["TrainingHours"] /
        (df_features["YearsWithCompany"] + 1)
    )

# 5. Long-distance employee flag
if "DistanceFromHome" in df_features.columns:
    df_features["LongDistanceEmployee"] = (
        df_features["DistanceFromHome"] > 20
    ).astype(int)

# 6. Overtime flag
if "DoesOvertime" in df_features.columns:
    df_features["OvertimeFlag"] = (
        df_features["DoesOvertime"]
        .astype(str)
        .str.lower()
        .map({"yes": 1, "no": 0})
        .fillna(0)
        .astype(int)
    )

print("Feature engineering completed!")

print("\nNew Dataset Shape:", df_features.shape)

print("\nNew Features:")
new_features = [
    col for col in df_features.columns
    if col not in df_clean.columns
]
print(new_features)

display(df_features.head())
```

```python
# Identify numerical and categorical columns

numerical_features = df_features.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

categorical_features = df_features.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Exclude 'LeftCompany' from categorical features as it is the target variable
if "LeftCompany" in categorical_features:
    categorical_features.remove("LeftCompany")

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)

print("\nTotal Numerical Features:", len(numerical_features))
print("Total Categorical Features:", len(categorical_features))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Numerical preprocessing
numerical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

# Combine numerical and categorical preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_pipeline, numerical_features),
        ("cat", categorical_pipeline, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
# Fix column names and recreate feature dataset

# Remove unwanted spaces from column names
df_features = df_clean.copy()
df_features.columns = df_features.columns.str.strip()

print("Columns in current dataset:")
print(df_features.columns.tolist())

# Check target column
if "LeftCompany" in df_features.columns:
    print("\n✅ LeftCompany column found!")
else:
    print("\n❌ LeftCompany column NOT found!")
```

```python
# ==========================================
# FINAL FEATURE ENGINEERING + PIPELINE SETUP
# ==========================================

# Start from cleaned dataset
df_features = df_clean.copy()

# Remove spaces from column names
df_features.columns = df_features.columns.str.strip()

# -------------------------------
# 1. Feature Engineering
# -------------------------------

# Annual Salary
if "MonthlySalary" in df_features.columns:
    df_features["AnnualSalary"] = df_features["MonthlySalary"] * 12

# Salary per year of experience
if "MonthlySalary" in df_features.columns and "YearsWithCompany" in df_features.columns:
    df_features["SalaryPerYearExperience"] = (
        df_features["MonthlySalary"] /
        (df_features["YearsWithCompany"] + 1)
    )

# Leave Rate
if "LeavesTaken" in df_features.columns and "YearsWithCompany" in df_features.columns:
    df_features["LeaveRate"] = (
        df_features["LeavesTaken"] /
        (df_features["YearsWithCompany"] + 1)
    )

# Training Intensity
if "TrainingHours" in df_features.columns and "YearsWithCompany" in df_features.columns:
    df_features["TrainingIntensity"] = (
        df_features["TrainingHours"] /
        (df_features["YearsWithCompany"] + 1)
    )

# Long Distance Employee
if "DistanceFromHome" in df_features.columns:
    df_features["LongDistanceEmployee"] = (
        df_features["DistanceFromHome"] > 20
    ).astype(int)

# Overtime Flag
if "DoesOvertime" in df_features.columns:
    df_features["OvertimeFlag"] = (
        df_features["DoesOvertime"]
        .astype(str)
        .str.lower()
        .map({"yes": 1, "no": 0})
        .fillna(0)
        .astype(int)
    )

# -------------------------------
# 2. Separate Target
# -------------------------------

target_column = "LeftCompany"

X = df_features.drop(columns=[target_column])
y = df_features[target_column]

# -------------------------------
# 3. Identify Columns
# -------------------------------

numerical_features = X.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)

# -------------------------------
# 4. Create Fresh Pipelines
# -------------------------------

numerical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_pipeline, numerical_features),
        ("cat", categorical_pipeline, categorical_features)
    ]
)

print("\nPreprocessing pipeline recreated successfully!")

# -------------------------------
# 5. Apply Pipeline
# -------------------------------

X_processed = preprocessor.fit_transform(X)

print("\nPreprocessing completed successfully!")
print("Original feature shape:", X.shape)
print("Processed feature shape:", X_processed.shape)
print("Target shape:", y.shape)
```

```python
# Get processed feature names
feature_names = preprocessor.get_feature_names_out()

# Create processed DataFrame
X_processed_df = pd.DataFrame(
    X_processed,
    columns=feature_names,
    index=df_features.index
)

# Add target column
X_processed_df["LeftCompany"] = y.values

print("Processed dataset created successfully!")
print("Final shape:", X_processed_df.shape)

display(X_processed_df.head())
```

```python
# Export processed dataset

output_file = "processed_employee_attrition_dataset.csv"

X_processed_df.to_csv(output_file, index=False)

print("Processed dataset exported successfully!")
print("File name:", output_file)

# Check exported file
print("Rows:", X_processed_df.shape[0])
print("Columns:", X_processed_df.shape[1])
```
